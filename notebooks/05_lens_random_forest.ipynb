# Use case — `Utility/RF.py`

Score every component pair within each source using the trained lens-feature random forest.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT)) if str(PROJECT_ROOT) not in sys.path else None

from Utility.RF import run_rf_inference
```

```python
INPUT_CSV = PROJECT_ROOT / "data" / "cleaned_lightcurves.csv"      # Preprocessed canonical light curves.
MODEL_PATH = PROJECT_ROOT / "models" / "lens22_rf.joblib"            # Saved RF bundle with feature order and configuration.
OUTPUT_CSV = PROJECT_ROOT / "results" / "rf_pair_predictions.csv"    # One probability per available pair.

BATCH_SIZE = 64              # Feature rows passed to predict_proba at once; larger can be faster.
KEEP_FAILED_PAIRS = True     # Preserve unscorable pairs with Proba=NaN for auditability.
MAX_PAIRS = None             # None scores all pairs; use a small integer for a pipeline test.
VERBOSE = True               # Print progress and the final scoring summary.

if not INPUT_CSV.exists():
    raise FileNotFoundError(f"Supply the cleaned canonical CSV first: {INPUT_CSV}")
if not MODEL_PATH.exists():
    raise FileNotFoundError(f"Saved RF model not found: {MODEL_PATH}")
```

```python
predictions = run_rf_inference(
    input_csv=INPUT_CSV,
    model_path=MODEL_PATH,
    output_csv=OUTPUT_CSV,
    batch_size=BATCH_SIZE,
    keep_failed_pairs=KEEP_FAILED_PAIRS,
    max_pairs=MAX_PAIRS,
    verbose=VERBOSE,
)
display(predictions.sort_values("Proba", ascending=False).head(20))
```
