# Use case — `Utility/Plot.py`

Inspect one or several components before classification or time-delay fitting.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT)) if str(PROJECT_ROOT) not in sys.path else None

from Utility.Plot import plot_lens_components_matplotlib
```

```python
INPUT_CSV = PROJECT_ROOT / "data" / "cleaned_lightcurves.csv"  # Preprocessed canonical table.
COMPONENT_IDS = ["COMPONENT_A", "COMPONENT_B"]                  # Replace with one or more exact Gaia component IDs.
Y_COLUMN = "flux_obs"                                           # Quantity shown on the vertical axis.
Y_ERROR_COLUMN = "flux_obs_error"                               # Measurement-error column; set None if absent.
Y_MODE = "raw"                                                   # "raw", "centered", "normalized", "zscore" or "minmax".
SHOW_POINTS = True                                               # Draw individual Gaia measurements.
SHOW_SMOOTH = False                                              # Add a display-only smooth trend; never used for fitting.
SMOOTH_WINDOW = 5                                                # Point window for the optional visual smooth.
MAX_POINTS_PER_CURVE = None                                      # Optional display downsampling; None shows every point.
TIMEZONE = "Europe/Paris"                                        # Timezone used when converting Julian dates for labels.

if not INPUT_CSV.exists():
    raise FileNotFoundError(f"Supply the cleaned canonical CSV first: {INPUT_CSV}")
```

```python
fig, ax = plot_lens_components_matplotlib(
    INPUT_CSV,
    lens_component_ids=COMPONENT_IDS,
    y_col=Y_COLUMN,
    yerr_col=Y_ERROR_COLUMN,
    y_mode=Y_MODE,
    show_points=SHOW_POINTS,
    show_smooth=SHOW_SMOOTH,
    smooth_window=SMOOTH_WINDOW,
    max_points_per_curve=MAX_POINTS_PER_CURVE,
    timezone=TIMEZONE,
)
fig
```
