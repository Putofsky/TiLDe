# Use case — `Utility/IdSep.py`

Group a flat Gaia table into `{source_id: {component_id: DataFrame}}`. Edit only the parameter cell, then run from top to bottom.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT)) if str(PROJECT_ROOT) not in sys.path else None

from Utility.IdSep import build_component_dict
```

## Parameters

Every parameter is documented inline. Keep Gaia identifiers as strings whenever the CSV may be opened in spreadsheet software.

```python
INPUT_CSV = PROJECT_ROOT / "data" / "cleaned_lightcurves.csv"  # Flat input table; replace with your file.
FIELD_COLUMN = "source_id"                                      # Groups measurements belonging to one lens system.
COMPONENT_COLUMN = "lensComponentSourceId"                     # Separates images/components within each system.
SORT_COLUMN = "epoch_obs_jd"                                    # Orders every light curve chronologically; use None to skip.
RESET_INDEX = True                                               # Gives each component a clean 0..N-1 row index.

if not INPUT_CSV.exists():
    raise FileNotFoundError(f"Supply the canonical CSV first: {INPUT_CSV}")
```

```python
components = build_component_dict(
    INPUT_CSV,
    field_col=FIELD_COLUMN,
    component_col=COMPONENT_COLUMN,
    sort_col=SORT_COLUMN,
    reset_index=RESET_INDEX,
)

print("Systems:", len(components))
first_source = next(iter(components))
print("First source:", first_source)
print("Components:", list(components[first_source]))
display(next(iter(components[first_source].values())).head())
```
